```python
from pathlib import Path
import pandas as pd
import numpy as np
import json
from jamo import h2j, j2hcj
import os

real_sen_morpheme_path = r"C:\Users\user\Downloads\수어 영상\1.Training\real_sen_morpheme\01"
real_word_morpheme_path = r"C:\Users\user\Downloads\수어 영상\1.Training\real_word_morpheme\01"
crowd_morpheme_path = r"C:\Users\user\Downloads\수어 영상\1.Training\crowd_morpheme\01"

morphemes = {'sen':{}, 'word':{}, 'crowd':{}}

def extractWord(morpheme_data: json):
    result = []
    
    isQuestion = False

    for idx in range(0, len(morpheme_data['data'])):
        word = morpheme_data['data'][idx]['attributes'][0]
        result.append(word['name'])
        
        if('attribute' in word):
            isQuestion = True
            result.append('?')
            
    # if(isQuestion):
    #     print(str(word['attribute']), ": ", result)

    return result

def extractWordToJamo(morpheme_data: json):
    #이중 모음 분리를 위한 맵핑
    COMPLEX_VOWELS = {
        'ㅘ': ['ㅗ', 'ㅏ'],
        'ㅙ': ['ㅗ', 'ㅐ'],
        'ㅚ': ['ㅗ', 'ㅣ'],
        'ㅝ': ['ㅜ', 'ㅓ'],
        'ㅞ': ['ㅜ', 'ㅔ'],
        'ㅟ': ['ㅜ', 'ㅣ'],
        'ㅢ': ['ㅡ', 'ㅣ'],
    }
    result = []
    
    #지문자는 단어 1개만 존재
    word = morpheme_data['data'][0]['attributes'][0]['name']
    word = j2hcj(h2j(word))
    
    for ch in word:
        if ch in COMPLEX_VOWELS:
            result.extend(COMPLEX_VOWELS[ch])
        else:
            result.append(ch)

    #print(result)
    
    return result

for senNum in range(1, 2001):
    for dire in ['U', 'D', 'F', 'L', 'R']:
        filename = f"NIA_SL_SEN{senNum:04d}_REAL01_{dire}_morpheme.json"
        morpheme_path = Path(f"{real_sen_morpheme_path}\{filename}")

        try:
            with open(morpheme_path, "r", encoding="utf-8") as f:
                morpheme = json.load(f)
        except Exception as e:
            raise Exception(f"json 파일 오류: {morpheme_path.name}")

        result = extractWord(morpheme)

        morphemes['sen'][f"{senNum}_{dire}"] = result

for wordNum in range(1, 3001):
    for dire in ['U', 'D', 'F', 'L', 'R']:
        filename = f"NIA_SL_WORD{wordNum:04d}_REAL01_{dire}_morpheme.json"
        morpheme_path = Path(f"{real_word_morpheme_path}\{filename}")

        try:
            with open(morpheme_path, "r", encoding="utf-8") as f:
                morpheme = json.load(f)
        except Exception as e:
            raise Exception(f"json 파일 오류: {morpheme_path.name}")

        result = extractWord(morpheme)

        morphemes['word'][f"{wordNum}_{dire}"] = result
        
for crowdNum in range(1, 1001):
    for dire in ['F']:
        filename = f"NIA_SL_FS{crowdNum:04d}_CROWD01_{dire}_morpheme.json"
        morpheme_path = Path(f"{crowd_morpheme_path}\{filename}")

        try:
            with open(morpheme_path, "r", encoding="utf-8") as f:
                morpheme = json.load(f)
        except Exception as e:
            raise Exception(f"json 파일 오류: {morpheme_path.name}")

        if(crowdNum <= 800):
            result = extractWordToJamo(morpheme)
        else:
            result = extractWord(morpheme)

        morphemes['crowd'][f"{crowdNum}_{dire}"] = result
```

```python
import cv2
import ast

#비디오, keypoint 프레임 수를 찾는 함수
def getVideoAndKeypointLength(video_path, keypoint_path):
    video_length = -1
    keypoint_length = -1
    
    if Path(video_path).exists():
        cap = cv2.VideoCapture(video_path)
        length = 0
        while True:
            ret, frame = cap.read()
            if not ret:
                break
            length += 1
        
        #video_length = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        video_length = length
        cap.release()
        
    if keypoint_path.exists():
        keypoints = pd.read_csv(keypoint_path)
        keypoint_length = len(keypoints)

    return video_length, keypoint_length
```

```python
data_ranges = pd.read_csv(r"C:\Users\user\Downloads\수어 영상\2.Validation\test_data_ranges.csv", encoding='utf-8-sig')

dataset = []

def appendDataset(dataset, data_range):
    for folderNum in range(data_range['folderStart'], data_range['folderEnd']):
        for num in range(1, data_range['number']):
            for dire in ast.literal_eval(data_range['dire']):
                filename = data_range['format'].format(num, folderNum, dire)
                
                other_path = Path(f"{data_range['video_path']}\{folderNum:02d}\{filename}__224.mp4")
                if(other_path.exists()):
                    os.rename(other_path, f"{data_range['video_path']}\{folderNum:02d}\{filename}_224.mp4")

                video_path = Path(f"{data_range['video_path']}\{folderNum:02d}\{filename}_224.mp4")
                keypoint_path = Path(f"{data_range['keypoint_path']}\{folderNum:02d}_merged\{filename}.csv")
                
                video_length, keypoint_length = getVideoAndKeypointLength(f"{data_range['video_path']}\{folderNum:02d}\{filename}_224.mp4", keypoint_path)

                dataset.append({
                    "video_path": str(video_path) if video_path.exists() else None,
                    "keypoint_path": str(keypoint_path) if keypoint_path.exists() else None,
                    "video_length": video_length,
                    "keypoint_length": keypoint_length,
                    "morpheme": morphemes[data_range['morpheme_name']][f'{num}_{dire}']
                })

for idx, data_range in data_ranges.iterrows():
    print(idx)
    appendDataset(dataset, data_range)
    
dataset = pd.DataFrame(dataset)
```

```python
dataset['match_ratio'] = (dataset[['video_length', 'keypoint_length']].min(axis=1) /
                            dataset[['video_length', 'keypoint_length']].max(axis=1)).round(3)
                         
dataset['morpheme_length'] = dataset['morpheme'].apply(
    #lambda x: len(ast.literal_eval(x))
    lambda x: len(x)
)

dataset['more_morp_length'] = dataset[['video_length', 'keypoint_length']].min(axis=1) / 2 < dataset['morpheme_length']

dataset.to_csv(r"C:\Users\user\Downloads\수어 영상\1.Training\testsetV4.csv", encoding='utf-8-sig', index=False)

#dataset[(dataset['match_ratio'] > 0.7)]

#dataset[(dataset['match_ratio'] <= 0.9) & (dataset['morpheme_length'] >= 5)]
```
